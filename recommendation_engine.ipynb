```python
# ======================================================
# RECOMMENDATION ENGINE MVP (Single File)
# ======================================================

# -----------------------------
# SECTION 1 : PRODUCT DATABASE
# -----------------------------

products = [
    {"id": 1, "name": "Nike Shoes", "category": "Shoes"},
    {"id": 2, "name": "Adidas Shoes", "category": "Shoes"},
    {"id": 3, "name": "MacBook", "category": "Electronics"},
    {"id": 4, "name": "Gaming Mouse", "category": "Electronics"},
    {"id": 5, "name": "Boat Earbuds", "category": "Electronics"},
    {"id": 6, "name": "Nike Hoodie", "category": "Clothing"},
    {"id": 7, "name": "Levis Jeans", "category": "Clothing"},
]

# ---------------------------------------------------
# SECTION 2 : USER ACTIVITY (SIMULATED WEBSITE DATA)
# ---------------------------------------------------
# 👇 Sirf is section ko change karo
# Future me ye data database se aayega

user_activity = [

    {"action": "wishlist", "product": 1},
    {"action": "wishlist", "product": 1},
    {"action": "wishlist", "product": 1},
    {"action": "purchase", "product": 7},

    {"action": "purchase", "product": 7},
    {"action": "purchase", "product": 7},

]

# ---------------------------------------------------
# SECTION 3 : ANALYTICS ENGINE
# ---------------------------------------------------

weights = {
    "view": 0,
    "wishlist": 3,
    "purchase": 21
}

category_score = {}

for event in user_activity:

    product = next(
        (p for p in products if p["id"] == event["product"]),
        None
    )

    if product is None:
        continue

    score = weights.get(event["action"], 0)

    category = product["category"]

    category_score[category] = category_score.get(category, 0) + score

# ---------------------------------------------------
# SECTION 4 : FIND FAVORITE CATEGORY
# ---------------------------------------------------

favorite_category = max(category_score, key=category_score.get)

# ---------------------------------------------------
# SECTION 5 : RECOMMENDATION ENGINE
# ---------------------------------------------------

recommendations = []

for product in products:

    if product["category"] == favorite_category:
        recommendations.append(product["name"])

# ---------------------------------------------------
# OUTPUT
# ---------------------------------------------------

print("=" * 50)
print("USER ACTIVITY")
print("=" * 50)

action_count = {}

for event in user_activity:
    action = event["action"]
    action_count[action] = action_count.get(action, 0) + 1

for action, count in action_count.items():
    print(f"{action.capitalize():10}: {count}")

print("\n" + "=" * 50)
print("CATEGORY ANALYTICS")
print("=" * 50)

for category, score in category_score.items():
    print(f"{category:15}: {score}")

print("\nFavorite Category :", favorite_category)

print("\n" + "=" * 50)
print("RECOMMENDED PRODUCTS")
print("=" * 50)

for item in recommendations:
    print("✔", item)

print("\n" + "=" * 50)
print("ENGINE FINISHED")
print("=" * 50)
```
